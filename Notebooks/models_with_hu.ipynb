```python
import cv2
import numpy as np
import pandas as pd
import os

# cv2.setLogLevel(cv2.LOG_LEVEL_ERROR)

def hu_moments_from_image(path):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    moments = cv2.moments(img)
    hu = cv2.HuMoments(moments).flatten()
    return hu

# Collect Hu moments for dataset
data = []
for label in ["can_hold", "cannot_hold"]:
    folder = os.path.join("/Users/gnbhavithran/Python_github/Bhavi_Yams/Tem_Sem_project/Dataset", label)
    for fname in os.listdir(folder):
        if fname.endswith(".tif"):
            path = os.path.join(folder, fname)
            hu = hu_moments_from_image(path)
            data.append({"label": label, **{f"hu{i+1}": hu[i] for i in range(7)}})

df = pd.DataFrame(data)
print(df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Pairplot of Hu moments
sns.pairplot(df, hue="label", vars=[f"hu{i+1}" for i in range(7)])
plt.show()

# Correlation heatmap
corr = df[[f"hu{i+1}" for i in range(7)]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation between Hu moments")
plt.show()
```
